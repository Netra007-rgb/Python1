## try except else finally
- The try block lets you test a block of code for errors
- The except block lets you handle the error
- The else block lets you execute code when there is not error
- The finally block lets you execute the code regardless of the result if the try and except block.

```python
def div(a,b):
    return a/b
try:
    print(div(5,0))
except:
    print('Exception Occured.')
else:
    print('No Exception')
finally:
    print('This is always printed')
```

```python
def div(a,b):
    return a/b
try:
    print(div(5,1))
except:
    print('Exception Occured.')
else:
    print('No Exception')
finally:
    print('This is always printed')
```

```python
# in thid code, we have given all the possible excpetion handling errors that can occur.
try:
    m=5
    f=open('abc.txt')
    print(f.read())
    print(m)
    print(5/0)
    l=[1,2,3]
    print(l[100])
except FileNotFoundError:
    print('File Not Found.')
except NameError:
    print('Variable Not Defined')
except ZeroDivisionError:
    print('Cannot divide by zero')
except IndexError:
    print('Index Out Of Range')
```
