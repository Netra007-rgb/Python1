## Tuples
1. we cannot perform any changes in tuple.
2. Tuple is read only version of list
3. Insertion order is preserved
4. Duplicates are allowed.
5. Heterogeneous objects are allowed.
6. We can differentiate duplicate objects bu using index.
7. Supports both +ve and -ve index.
8. represent tuple elements within parenthesis (()) with comma separator
9. parenthesis are optional but recommended to use.

```python
t=tuple(range(10,20,2))
print(t)
```

```python
t=(10)
print(type(t))
q=(10,)
print(type(q))
```

### Accessing the elements of tuple
1. by using index
2. by using slicing operator

```python
t=(10,20,30,40,50,60)
print(t[0])
print(t[1])
print(t[2])
print(t[-1])
print(t[-2])
#print(t[10]) it gives index out of bound error.
```

```python
t=(10,20,30,40,50,60)
print(t[2:5])
print(t[2:100])
print(t[2:])
print(t[::2])
print(t[:5])
```

### Mathematical Operation for tuple
1. '+' -> concatenation
2. '*' -> repeatition

```python
t1=(10,20,30)
t2=(40,50,60)
t=t1+t2
print(t)
t=t1*3
print(t)
```

### Tuple Packing And Unpacking

```python
t=(10,20,30,40)
a,b,c,d=t
print(a,b,c,d)
```

### IMP functions of tuple

```python
# 1.len()
t=(10,20,30,40)
print(len(t))
```

```python
# 2. count()
t=(10,20,10,30,40,10)
print(t.count(10))
```

```python
# 3. index()
t=(10,20,10,30,40)
print(t.index(10))
print(t.index(40))
```

```python
# 4. sorted()
t=(10,20,10,40,30)
print(sorted(t))
```
