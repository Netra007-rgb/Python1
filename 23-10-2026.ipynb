## Lambda Function
-> it is a function without any name, anonymous function or lambda function.
Syntax:- lambda n:n*n

```python
s=lambda n:n*n
print(s(4))
print(s(5))
```

```python
l=['apple','banana','cherry','orange']
l.sort(key=lambda x:x[-1])
print(l)
```

```python
s=lambda a,b:a+b
print(s(10,20))
```

```python
# map function.
# for every element present in the given sequence, apply some functionality and generate new element with required modification.
# Syntax:- map(function,sequence)
l=[1,2,3,4,5]
def double(x):
    return 2*x
l1=list(map(double,l))
print(l1)
```

```python
l=[1,2,3,4,5]
l1=list(map(lambda x:2*x,l))
print(l1)
```

```python
l1=[1,2,3,4,5]
l2=[4,5,6,7,8]
l3=list(map(lambda x,y:x*y,l1,l2))
print(l3)
```

```python
# reduce function.
# reduce the sequence of elements into a single element by applying specified function.
# Syntax:- reduce(function,sequence)
# reduce() function present in functools module so we need to import the module to use it.
from functools import *
l=[10,20,30,40,50]
sum=reduce(lambda x,y:x+y,l)
print(sum)
mul=reduce(lambda x,y:x*y,l)
print(mul)
```

```python
# filter function()
# we can use filter() function to filter value from given sequence based on same condition.
# Syntax:- filter(functon,sequence)
# where function argument is responsible to perform conditional check and sequence can be list or tuple or string.
def iseven(x):
    if x%2==0:
        return True
    else:
        return False
l=[5,10,15,20,25,30,35,40,45]
l1=list(filter(iseven,l))
print(l1)
```

```python
l=[5,10,15,20,25,30,35,40,45]
l1=list(filter(lambda x:x%2==1,l))
print(l1)
```

```python
# wap to convert the list of strings to uppercase using map function.
l=['a','b','c','d']
def u(x):
    return x.upper()
s=list(map(u,l))
print(s)
```

```python
# wap to filter the words with length>5.
l=['apple','banana','cherry','orange','plum']
s=list(filter(lambda x:len(x)>5,l))
print(s)
```

```python
# wap to find the numbers divisible by 19 or 13 from a list of numbers using lambda function.
n=[19,37,38,36,47,52,65,76,88,85]
a=lambda x:[num for num in x if num%19==0 or num%13==0]
print(a(n))
```
