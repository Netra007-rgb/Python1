```python
def fun(n1,*s): # we have fixed the value if n1 but we didn't fixed the length of variable s.
    print(n1)
    for s1 in s:
        print(s1)
fun(10)
fun(10,20,30)
fun(10,"A","B",20,30)
```

```python
def fun(*s,n1): # we have fixed the value if n1 but we didn't fixed the length of variable s.
    for s1 in s:
        print(s1)
    print(n1)
fun(10) # n1=0
fun(10,20,30) #n1=30
fun(10,"A","B",20,30) # n1=30  # but if we assign the number to n1 while calling the function then error will not be interpreted.
```

```python
def fun(**n): # *->args and **->kwargs
    print(n)
    for k,v in n.items():
        print(k,"=",v)
fun(m1=100,m2=200,m3=300,m4=400)
fun(m0=1,name="Arman",marks=80)
```

### Nested Function

```python
def outer():
    print('This is Outer Function')
    def inner():
        print('This is inner function')
    return inner() # after outer is called it will call the inner function and it will return inner function.
outer()
```

```python
def happy(num):
    temp=num
    while(num!=1 and num!=4):
        sum=0
        s=str(num)
        for i in s:
            sum+=int(i)**2
            num=sum
    if num==1:
        print('Happy number')
    else:
        print('Not a happy number')
n=int(input('Enter your number to check whether it is a happy number or not.'))
happy(n)
```

```python
# Harshad Number
def harshad(num):
    sum=0
    temp=n
    while(num>0):
        rem=num%10
        sum=sum+rem
        num=num//10
    if temp%sum==0:
        print('harshad number')
    else:
        print('not a harshad number')
n=int(input('Enter your number'))
harshad(n)
```

```python
# wap to print all chronic numbers between 1 to 100.
# num=n(n+1) i.e:- 6=2(2+1)
def chronic():
    for i in range (100):
        for j in range(i):
            if (j*(j+1)==i):
                print(i,end=" ")
                break
chronic()
```
