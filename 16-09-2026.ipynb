```python
# wap to find the sum of n numbers. n is to be entered by users.
n=int(input('Enter your number.'))
sum=0
for i in range (1,n+1):
    sum=sum+i
print('sum=',sum)
```

```python
# wap to read 3 numbers from user and check how many numbers available between a and b are divisible by c.
a=int(input('Enter the 1st number:'))
b=int(input('Enter the 2nd number:'))
c=int(input('Enter the 3rd number:'))
count=0
for i in range (a+1,b):
    if(i%c==0):
        count+=1
print(count)
```

```python
# wap to find the factorial of a number entered by user without use of math library.
num=int(input('Enter your num:'))
fact=1
for i in range(1,num+1):
    if num==0:
        fact=1
    else:
        fact=fact*i
print('factorial=',fact)
```

```python
# wap to display fibonacci sequence upto n term, n will be entered by user.
num=int(input('Enter your number:'))
n1=0
n2=1
if n==1:
    print('Fibonacci series:',n1)
else:
    print('Fibonacci series:')
    print(n1)
    print(n2)
    for i in range (n-1):
        nth=n1+n2
        print(nth)
        n1=n2
        n2=nth
```

```python
# wap to take 10 numbers from users and print their sum and average.
sum=0
avg=0
for i in range(0,10):
    num=int(input('Enter num:'+str(i+1)))
    sum=sum+num
avg=sum/10
print("Sum=",sum)
print("Avg=",avg)
```
