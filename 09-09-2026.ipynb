### Operators

```python
# 1.Arithmetic Operator.
a=12
b=5
print("a+b= ",a+b)
print("a-b= ",a-b)
print("a*b= ",a*b)
print("a/b= ",a/b) # in division by default ans will be in float.
print("a%b= ",a%b)
print("a//b= ",a//b) # this is flow division and if both identifiers are integer then ans will be in integer.
print("a**b= ",a**b) # power i.e:-12^5
```

```python
# 2.Comparision Operators(Relational)
a=10
b=20
print("a>b-> ",a>b)
print("a<b-> ",a<b)
print("a>=b-> ",a>=b)
print("a<=b-> ",a<=b)
print("a==b-> ",a==b)
print("a!=b-> ",a!=b) 
```

```python
a='arman'
b='Aryan'
print(a>b) # checks according to ascii values.
print(a<b)
print(10<20<30)
print(10<20<30>40)
print(False>True) # 0>1 is false.
```

```python
# 3.Logical Operators.
#->for boolean type behaviour
# and->if both arguments are True then only result is True
# or->if atleast one argument is True then result is True
# not->complement

#->for non-boolean type behaviour
# x and y->if x is evaluate to false return x otherwise return y
# x or y-> if x evaluates to True then result is x otherwise y
# not x->if x is evaluates to False then result is True otherwise False
```
