## Set
-> mutuable
-> represent group of unique values as a single entity
-> duplicates are not allowed
-> insertion order not preserved but we can sort the elements 
-> indexing and slicing not allowed for set
-> heterogeneous objects allowed
-> we can represent set elements within curly brackets with common seperator 
-> we can apply mathematical operations like union,intersection,difference etc on set objects

```python
# Creation Of Set
s={10,20,30,40}
print(s)
print(type(s))
```

```python
s=set()
print(type(s))
```

```python
s=[10,20,30,40,10,20,30,40]
x=set(s)
print(x)
s=list(x)
print(s)
```

```python
s=set(range(10))
print(s)
```

### IMP Functions Of Set

```python
# 1. len()
s={1,2,3,4,1,2,3,4,5,6}
print(len(s)) # without duplicates
```

### Mathematical Operations

```python
# 1. union()
x={10,20,30,40}
y={30,40,50,60}
print(x.union(y)) # or print(x|y)
print(x)
print(y)
```

```python
# 2. intersection()
x={10,20,30,40}
y={30,40,50,60}
print(x.intersection(y)) # or print(x&y)
```

```python
# 3. difference()
x={10,20,30,40}
y={30,40,50,60}
print(x.difference(y)) # or print(x-y)
print(y.difference(x)) # or print(y-x)
```

```python
# 4. symmetric_difference() -> elements that are not common and they are merged by union.
x={10,20,30,40}
y={30,40,50,60}
print(x.symmetric_difference(y)) # or print(x^y) 
```

```python
# 5. issubset()
x={'a','b','c'}
y={'f','e','c','b','a'}
z=x.issubset(y)
p=y.issubset(x)
print(z)
print(p)
```

```python
# 6. issuperset() _. just opp. of issubset()
x={'a','b','c'}
y={'f','e','c','b','a'}
z=x.issuperset(y)
p=y.issuperset(x)
print(z)
print(p)
```

```python
# Memebership Operator in Set.
s=set('Arman')
print(s)
print('x' in s)
print('a' in s)
print('y' not in s)
```

```python
# 7. add() -> cannot add by index.
s={10,20,30}
s.add(40)
s.add(50)
print(s)
```

```python
# 8. update()
s={10,20,30}
l=[40,50,60,10]
s.update(l,range(5),range(10,20))
print(s)
```

```python
# 9. copy()
s={10,20,30}
x=s.copy()
print(x)
```

```python
# 10. pop()
s={10,20,30,40}
print(s)
print(s.pop())
print(s)
```
