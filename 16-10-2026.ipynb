## Dictionary
-> group of objects as key value pairs i.e:- rollno-name primeno-address
-> Duplicate keys are not allowed but values can be duplicated
-> Heterogeneous objects allowed for both keys and values
-> Insertion order is not preserved
-> Indexing and slicing concepts are not applicable

### Creation Of Dictionary
d={} or d=dict()

```python
d={1:'Arman',2:'Dev',3:'Dhairya'}
d[1]='Karma' #d[key]=new_value
print(d)
print(d[3]) # value can be printed by using key.
#print(d[10]) key error is generated
d[4]='Sumit'
d[5]='John'
print(d) # it wil be updated.
```

```python
# Deleting the elements of dictionary.
# syntax:-del d[key]
del d[2]
del d[4]
#del d[10] generates an error if key not present.
print(d)
```

```python
d.clear()
print(d)
```

```python
# Dictionary Comprehension
square={x:x*x for x in range(1,6)}
print(square)
```

### IMP functions of Dictionary

```python
# 1. dict()
l=[(10,'Arman'),(20,'Aryan'),(30,'Dhairya')]
d=dict(l)
print(d)
```

```python
# 2. len()
d={10: 'Arman', 20: 'Aryan', 30: 'Dhairya'}
print(len(d)) # acc to keys
```

```python
# 3. clear()
d.clear()
print(d)
```

```python
# 4. copy()
d1=d.copy()
print(d1)
```

```python
# 5. update()
d1={1:5,2:30,3:40}
d2={3:50,4:60,5:70}
d1.update(d2)
print(d1)
```

```python
# 6. get()
d={10:'Arman',20:'Aryan',30:'Dhairya'}
print(d[10])
print(d.get(10))
print(d.get(40))
```

```python
print(d.get(10,'Guest'))
print(d.get(40,'Guest')) # default value if the key isn't found then it returns the default value.
```

```python
# 7. popitem()
d={10:50,20:80,30:100}
print(d.popitem()) # removes a random value from the dictionary.
```
