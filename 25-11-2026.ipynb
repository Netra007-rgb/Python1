```python
f=open('mbox-short.txt')
l=[]
for line in f:
    if 'X-DSPAM-Confidence:' not in line:
        continue
    x=line.split(":")
    l.append(x[1].strip())
print(l)
sum=0
for i in l:
    sum+=float(i)
print('Average Value for Span Confidence:',sum/len(l))
f.close()
```

```python
f=open('mbox-short.txt')
counts={}
for line in f:
    if line[:5] == 'From:':
        words=line.split()
        emails=words[1].strip()
        if emails in counts:
            counts[emails]+=1
        else:
            counts[emails]=1
print(counts) 
f.close()
```
