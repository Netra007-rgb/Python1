### Accessing Characters using Slicing Operator
Syntax:- s[begin index : end index : step]

```python
s='Python is very easy'
s[1:7:1] # starts from 1 and stops right before 7th character
```

```python
s='Python is very easy'
s[1:7] # by default it increments by 1 step
```

```python
s='Python is very easy'
s[1:7:2]
```

```python
s='Python is very easy'
s[:7] # consider 0 if starting is not given
```

```python
s='Python is very easy'
s[9:] # starts from 9 and ending is not given so it goes uptill end.
```

```python
s='Python is very easy'
s[9::]
```

```python
s='Python is very easy'
s[::] # by default starts from beginning and goes uptill end and increments by 1.
```

```python
s='Python is very easy'
s[::-1] # prints reverse.
```

```python
s='Arman'
s[-4::]
```

```python
s='Arman'
s[-4::-1]
```

```python
s='Arman'
s[-4:1:-1]
```

```python
# isspace()-> checks whether space is present or not.
s="Python is very easy"
count=0
for i in s:
    if i.isspace():
        count+=1
print(count)
```

```python
# wap to count all the letters,digits and special symbols from a given string
s=input('Enter your String')
lcount=0
dcount=0
scount=0
for i in s:
    if i.isdigit():
        dcount+=1
    elif i.isalpha():
        lcount+=1
    else:
        scount+=1
print('letters,digits,special symbols:',lcount,dcount,scount)
```

```python
s="Py$@th173$54on389"
sum=0
for i in s:
    if i.isdigit():
        sum+=int(i)
print('Sum:',sum)
avg=sum/8
print('Average:',avg)
```
