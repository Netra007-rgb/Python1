## Deleting Directory Or Files 
### os.remove()
- Used to remove or delete a file. This method can not remove or delete a directory. If specified path is directory then OSError will be raised by the 
method.

### os.rmdir()
- This method is used to remove or delete an empty directory. OSError will be raised if the specified path is not an empty directory.

```python
import os
os.remove('function.txt')
```

## datetime module
- date: represents a calender date
- time: represents a time of a day
- datetime: combines both date and time information

```python
from datetime import date
d=date(2002,12,31)
print(d)
```

```python
from datetime import time
t=time(12,10,0)
print(t)
```

```python
from datetime import datetime
now=datetime.now()
print(now)
```
