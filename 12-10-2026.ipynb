```python
# enumerate()
# enumerate(iterable,start=0) ->here start denotes the start point of the characters.
l1=('apple','banana','cherry')
s1='geek'
obj1=list(enumerate(l1))
print(obj1)
obj2=list(enumerate(s1,start=21))
print(obj2)
for i,j in enumerate(l1):
    print(i,j)
```
