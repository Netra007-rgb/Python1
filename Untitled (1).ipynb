```python
# 1. function with no parameters and no return type
def printline():
    s=input("Enter Name:")
    print(s)
printline() # calling the function
    
```

```python
# 2. function with parameters and no return type.
def printline(s):
    print(s)
s=input("Enter Name")
printline(s) # we gave paramter
```

```python
# 3. function with parameters and with return type
# if function doesn't return and if we print then it will show None.
def printline(s):
    return s
s=input('Enter Name')
t=printline(s) # we have to store the function in variable with return type.
print(t)
```

```python
# 4. function with no parameters and with return type.
def printline():
    s=input('Enter Name')
    return s
t=printline()
print(t)
```

### Types of argument
1. Positional argument
2. Keyword argument
3. Default argument
4. Variable length argument

```python
# 1. Positional Arguments
def sub(a,b):
    print(a-b)
sub(50,30)
sub(30,50)
```

```python
# 2. Keyword Arguments
def wish(name,msg):
    print("Hello",name,msg)
wish(name="Arman",msg="Good morning") # also if we change the sequence then also it will perfectly work becaus ewe are assingning the variable.
```

```python
def wish(name,msg):
    print("Hello",name,msg)
wish(name="Arman","Good morning")
```

```python
def wish(name,msg):
    print("Hello",name,msg)
wish("Arman",msg="Good morning") # it doesn't give error because in the 1st we didn't gave keyword its fine but if we give the keyword in 1st then 
# we have to compulsory go with the 1st parameter situation.
```

```python
# 3. Default Arguments
def wish(name="Guest"):
    print("Hello",name,"Good Morning")
wish("arman") # if we give argument then it will work accordingly otherwise the default argument will be executed.
wish()
```

```python
def wish(name,msg="Good Morning"): # here name is considered as positional argument and it won't give error like the keyword argument
    print("Hello",name,msg)
wish("Arman")
```

```python
def wish(name="Guest",msg):
    print("Hello",name,msg)
wish("Arman") # we haven't given the 2nd parameter.
```

```python
# 4. Variable length Argument
def fun(*n): # we can give number of arguments that we want.
    total=0
    for i in n:
        total=total+n
        print('Sum:',total)
fun(10,20,30)
fun(20,30)
```
