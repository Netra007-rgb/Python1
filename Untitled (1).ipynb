### File Access Modes
1. read mode(r)
2. read and write(r+)
3. write(w)
4. write and read(w+)
5. append mode(a)
6. append and read(a+)

```python
f=open('test.txt','r') # file name,mode
print(f.read())
print(f.closed)  # file is open.
f.close() # we have to close the file.
print(f.closed) # file closed.
```

```python
with open('test.txt','r') as f:
    print(f.read())
print('Hello') # file is automatically closed when we exit the intentation.
print(f.closed) # file is closed
```

```python
f=open('demo.txt') # if not given then by default the mode is read.
print(f.read()) # if the given file doesn't exist.
```

```python
f=open('test.txt','r')
f.write('XYZ') # cannot write in the given read mode.
```

```python
t=open('utf8_demo.txt','r',encoding='utf8') # we have to write to encode it to utf8 version because it is not suppported by any file so it is advisable to write utf8.
print(t.read())
```

```python
# read and write(r+)
f=open('test.txt','r+')
print(f.read()) # if we do not use read and use write then the cursor is at the starting position so it starts to overwrite.
f.write('New Data\n')
f.write('My Data')
f.close()
```

```python
print(f.seek(value)) # if we want the cursor at some specific position.
print(f.tell()) # tells the current position of the cursor.
```

```python
# write mode(w) -> old data is erased completely.
f=open('test.txt','w')
f.write('This is test message')
f.close()
```

```python
# append only (a) -> its added at the end.
f=open('test.txt','a')
f.write("New Data\n")
f.close()
```

```python
# append and read(a+)
f=open('test.txt','a+')
f.write('New Data\n')
f.seek(0)
print(f.read())
f.close()
```

```python
# Reading the files.
# 1. read()
f=open('test.txt','r')
print(f.read(5))
print(f.read(10))
print(f.read(7))
print(f.read())
f.close()
```

```python
# 2. readline()
f=open('test.txt','r')
print(f.readline(20)) # 1st line but is not printed.
print(f.readline(3)) # 2nd line
print(f.readline()) # 3rd line is printed.
f.close()
```

```python
# 3. readlines()
f=open('test.txt','r')
x=f.readlines()
print(x[2])
f.close()
```
