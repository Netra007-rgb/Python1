```python
# find() -> returns the index of first occurrence of the given substring if it isn't available then we will get -1.
s='Learning Python is very easy'
print(s.find('P'))
print(s.find('e')) # gives according to the first occurence.
print(s.find('Python')) # it checks for the word occurence.
print(s.find('Java'))
print(s.find('e',2)) # for finding the occurence other that we want we place the starting index and the ending index for searching.
```

```python
# split() -> splits the string by given symbol
s='University LJIET'
l=s.split()
print(l) # splits by space.
q='09-10-2026'
p=q.split('-') # splits from '-'
print(p)
s='LJ University LJIET LJ UNIVERSITY'
l=s.split(maxsplit=2) # splits uptill 2 words.
print(l)
s="LJ    UNIVERSITY   "
l=s.split()
print(l)
s='LJ LJ LJIET'
l=s.split('LJ')
print(l)
s='LJ LJ LJIET'
l=s.split('LJ' ,maxsplit=1)
print(l)
```

```python
# translate() with maketrans function.
import string
print(string.punctuation)
t='Py$@thon is very ea$sy'
x=t.maketrans("t","a",'$@') # makes the special character given in the argument none.
print(x)
print(t.translate(x))
t='Hello Sam!'
x=t.maketrans('S','P')
print(t.translate(x))
x='mSa'
y='eJo'
z=t.maketrans(x,y)
print(t.translate(z))
```

```python
# wap to display all the positions of a substring in a given string.
string=input('Enter your string.')
subs=input('Enter the string.')
flag=False
pos=-1
n=len(string)
c=0
while True:
    pos=string.find(subs,pos+1,n)
    if pos==-1:
        break
    c=c+1
    print('Found at positions:',pos)
    flag=True
    if flag==False:
        print('Not Found')
    print('The number of occurence:',c)
```

```python
# wap to replace each special symbols in a string with #
import string
s=input('Enter your string.')
x=s.maketrans(string.punctuation,'#'*32)
print(string.translate(x))
```

```python
# wap to merge the characters of 2 string into a single string by taking characters alternatively.
s1=input('Enter your 1st string')
s2=input('Enter your 2nd string')
output=""
i,j=0,0
while i<len(s1) or j<len(s2):
    output=output+s1[i]
    i=i+1
    output=output+s2[j]
    j=j+1
print(output)
```
